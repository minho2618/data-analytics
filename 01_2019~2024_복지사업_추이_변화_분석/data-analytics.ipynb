```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn\
    as sns
import platform

# 1. 한글 폰트 설정 (환경에 따라 폰트 깨짐 방지)
if platform.system() == 'Windows':
    plt.rc('font', family='Malgun Gothic')
elif platform.system() == 'Darwin':
    plt.rc('font', family='AppleGothic')
else:
    plt.rc('font', family='NanumGothic')
plt.rcParams['axes.unicode_minus'] = False

# 2. 데이터 로드 및 전처리
files = [
    '복지사업 월별, 시도별 수급권자 현황(2019년도).csv',
    '복지사업 월별, 시도별 수급권자 현황(2020년도).csv',
    '1_1.복지사업 월별, 시도별 수급권자 현황(2021년도).csv',
    '한국사회보장정보원_복지사업 월별 시도별 수급권자 현황_20221231.csv',
    '1-2_한국사회보장정보원_복지사업 월별 시도별 수급권자 현황_20231231.csv',
    '한국사회보장정보원_복지사업 월별 시도별 수급권자 현황_20241231.csv'
]

df_list = []
for file in files:
    try:
        # 공공데이터의 보편적 인코딩(cp949) 우선 시도
        df_temp = pd.read_csv(file, encoding='cp949')
        df_list.append(df_temp)
    except UnicodeDecodeError:
        try:
            df_temp = pd.read_csv(file, encoding='euc-kr')
            df_list.append(df_temp)
        except Exception as e:
            pass

df = pd.concat(df_list, ignore_index=True)

# 컬럼명 및 텍스트 공백 제거를 통한 데이터 정제
df.columns = df.columns.str.strip()
if '사업명' in df.columns:
    df['사업명'] = df['사업명'].str.replace(' ', '')

# 기준년월에서 연도(4자리) 추출
df['기준년월'] = df['기준년월'].astype(str).str.replace('-', '')
df['연도'] = df['기준년월'].str[:4]

# 수급권자수 자료형 변환 (문자열 내 쉼표 제거 후 실수 변환)
if df['수급권자수'].dtype == 'object':
    df['수급권자수'] = df['수급권자수'].astype(str).str.replace(',', '').astype(float)
else:
    df['수급권자수'] = df['수급권자수'].astype(float)

# 연도별, 사업별 연평균 수급권자수 산출
df_sum_by_month = df.groupby(['연도', '기준년월', '사업명'])['수급권자수'].sum().reset_index()
yearly_df = df_sum_by_month.groupby(['연도', '사업명'])['수급권자수'].mean().reset_index()

# 3. 산출물 A: 복지사업별 연도별 변화 분석 (Line Chart)
plt.figure(figsize=(12, 6))
sns.lineplot(data=yearly_df, x='연도', y='수급권자수', hue='사업명', marker='o')
plt.title('2019~2024 복지사업별 수급권자수 추이')
plt.xlabel('연도')
plt.ylabel('평균 수급권자수 (명)')
plt.legend(bbox_to_anchor=(1.05, 1), loc='upper left')
plt.tight_layout()
plt.show()

# 4. 산출물 B: 2024년 사업별 수급권자 규모 순위 그래프
df_2024 = yearly_df[yearly_df['연도'] == '2024'].sort_values(by='수급권자수', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(data=df_2024, x='수급권자수', y='사업명', palette='viridis')
plt.title('2024년 복지사업별 수급권자 규모 순위')
plt.xlabel('평균 수급권자수 (명)')
plt.ylabel('사업명')
plt.tight_layout()
plt.show()

# 5. 산출물 C: 2019~2024 사업별 증가율 비교 그래프
df_2019 = yearly_df[yearly_df['연도'] == '2019']
growth_df = pd.merge(df_2019[['사업명', '수급권자수']], df_2024[['사업명', '수급권자수']], on='사업명', suffixes=('_2019', '_2024'))

# 증가율 계산식 적용
growth_df['증가율(%)'] = ((growth_df['수급권자수_2024'] - growth_df['수급권자수_2019']) / growth_df['수급권자수_2019']) * 100
growth_df = growth_df.sort_values(by='증가율(%)', ascending=False)

plt.figure(figsize=(10, 6))
sns.barplot(data=growth_df, x='증가율(%)', y='사업명', palette='coolwarm')
plt.title('2019~2024 복지사업별 수급권자 증가율 비교')
plt.xlabel('증가율 (%)')
plt.ylabel('사업명')
plt.tight_layout()
plt.show()
```
